```python
import pandas as pd
import numpy as np

data = {
    "EmpID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Arun", "Priya", "Kumar", "Divya", "Rahul",
        "Meena", "Sathish", "Nila", "Vijay", "Kavya"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "IT",
        "HR", "Finance", "IT", "Sales", "Sales"
    ],
    "Age": [25, 29, 32, 27, 24, 35, 31, 28, 30, 26],
    "Salary": [
        35000, 42000, 58000, 39000, 30000,
        62000, 51000, 47000, 44000, 36000
    ],
    "Experience": [2, 5, 8, 3, 1, 10, 7, 4, 6, 2],
    "City": [
        "Chennai", "Madurai", "Chennai", "Coimbatore", "Salem",
        "Chennai", "Madurai", "Coimbatore", "Chennai", "Salem"
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Q1 Loading and Inspecting the data
#Creating a sample data frame from the given dataset and Displaying the first 5 rows and last 3 rows,the number of rows and coloumns and the coloumn datatype  
print("First 5 rows:")
print(df.head(5))
print("\n last 3 rows:")
print(df.tail(3))
print("\n Number of rows and columns:")
print(df.shape)
print("\n Datatype:")
print(df.dtypes)
```

```python
#Q2 filerting employee records
#Display the details of employees who work in the IT department and have a salary greater than 40,000.
result=df[
    (df["Department"]== "IT") & (df["Salary"]>40000)
]
print(result)
```

```python
#Q3 Sorting the data 
#Displaying all employees sorted by salary from highest to lowest. Showing only Name, Department, and Salary.
result=df.sort_values(
    by="Salary",
    ascending=False
)[["Name","Department","Salary"]]
print(result.to_string(index=False))
```

```python
#Q4 Calculating summary statistics 
#Finding the mean, maximum, minimum, and median salary. Also finding the average age of all employees.
print("Mean Salary of the Employees =",df["Salary"].mean())
print("Maximum Salary of the Employees =",df["Salary"].max())
print("Minimun Salary of the Employees =",df["Salary"].min())
print("Median Salary of the Employees =",df["Salary"].median())
print("Average Age of Employees =",df["Age"].mean())
```

```python
#Q5 Group by Department
#For each department, calculating the number of employees, average salary, and highest salary. Sort the result by average salary in descending order. 
result=df.groupby("Department").agg(
    Employeecount=("EmpID","count"),
    AverageSalary=("Salary","mean"),
    HighestSalary=("Salary","max")
)
result=result.sort_values(
   by="AverageSalary",
   ascending=False
)
print(result)
```

```python
#Q6 Creating a new column 
#Creating a new column named AnnualSalary by multiplying Salary by 12 and Displaying Name, Salary, and AnnualSalary.
df["AnnualSalary"]=df["Salary"]*12
print(
    df[["Name","Salary","AnnualSalary"]])
```

```python
#Q7  Findind the  missing and duplicate values 
#Make a copy of the DataFrame, set one employee's City value to missing, and add a duplicate row. Use Pandas to identify missing values, fill the missing City with 'Unknown', and remove duplicate rows. 

df1=df.copy()
df1.loc[df1["EmpID"]==101,"City"] = np.nan
df1=pd.concat(
    [df1,df1.iloc[[1]]],
    ignore_index=True
)
print("Missing values : ")
print(df1.isnull().sum())
print()
print("Duplicate rows : ")
print(df1[df1.duplicated()])

df1["City"]= df1["City"].fillna("Unknown")
df1=df1.drop_duplicates()
print()
print("Cleaned data :")
print()
print("Missing values:",df1.isnull().sum().sum())
print("\n Duplicate rows",df1.duplicated().sum())
print("\n Total rows:",len(df1))




```

```python
#Q8 using conditional logics
'''Create a column named SalaryLevel: 'High' for salary greater than or equal to 50,000, 'Medium' for salary from 40,000 to 49,999, and
'Low' for salary below 40,000. Display Name, Salary, and SalaryLevel.'''
def salary_level(Salary):
    if Salary >=50000:
        return "High"
    elif Salary >=40000:
        return "Medium"
    else:
        return "Low"

df["SalaryLevel"]= df["Salary"].apply(salary_level)
print(df[["Name","Salary","SalaryLevel"]])

```

```python
#Q9  Work with a second DataFrame 
'''Create a marks DataFrame with columns EmpID and TrainingScore, containing scores for EmpID 101, 103, 105, 108, and 110.
Merge it with the employee DataFrame using EmpID. Display all employees and their scores, including employees without a score.''' 

marks = pd.DataFrame({
    "EmpID": [101, 103, 105, 108, 110],
    "TrainingScore": [87, 99, 76, 83, 90]
})

result = pd.merge(
    df,
    marks,
    on="EmpID",
    how="left"
)

print(
    result[["EmpID", "Name", "TrainingScore"]]
    .to_string(index=False)
)
```

```python
#Q10 Answer a business question 
'''Findind the department with the highest average salary. Then displaying the employees whose salary is above 
the overall average salary '''
dept_avg=df.groupby("Department")["Salary"].mean()
highest_dept=dept_avg.idxmax()
highest_avg=dept_avg.max()
print("Department with Highest Average Salary")
print(highest_dept ,"-",highest_avg)
overall_average=df["Salary"].mean()
print("\nOverall Average Salary:")
print(overall_average)
above_average=df[df["Salary"] > overall_average]
print("\nList of the Employees earning above the overall average salary:")
print(
    above_average[["Name","Department","Salary"]].to_string(index=False)
)


```
